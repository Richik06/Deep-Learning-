### Matrix Multiplication

Two main ways of performing multipliaction in neural networks and deep learning:

1. Element-wise multiplication
2. Matrix multiplication(dot product)

There are two main rules that performing matrix multiplication needs to satisfy

1. The inner dimension must match
* (3,2) @ (3,2) wont work
* (2,3) @ (3,2) will work
* (3,2) @ (2,3) will work

2. The resulting matrix has the shape of the outer dimension
* (2,3) @ (3,2) -> (2,2)

```python
torch.matmul(torch.rand(3,2),torch.rand(2,3))
```

### Reshaping, stackcing , squezzing and unsquezzing tensors

* Reshaping- reshapes an input tensor to a defined shape
* View- Return a view of an input tensor of certain shape but keep the same memory as the original tensor
* Stacking - combines multiple tensors on top of each other (vstack) or side by side (hstack)
* Squeeze- removes all 1 dimension from a tensor
* Unsqeeze- add a 1 dimension to a target tensor
* Permute- Return a view of the input with dimension permutted(swapped) in a certain way

```python
# Create tensor
import torch
x=torch.arange(1.,10.)
x,x.shape
```

```python
# Add an extra dimension
x_reshaped=x.reshape(1,9)
x_reshaped,x_reshaped.shape
```

```python
#  Change the view
z=x.view(1,9)
z,z.shape
```

```python
# Changing z changes x ( because a view of a tensor shares the same memory as the original tensor)
z[:,0]=5
z,x
```

```python
# stack tensors on top of each other
x_stacked=torch.stack([x,x,x,x], dim=1)
x_stacked
```

```python
# torch.squeeze
print(f"Previous tensor: {x_reshaped}")
print(f"Previous shape: {x_reshaped.shape}")

# Remove extra dimension from x_reshaped
x_squeezed=x_reshaped.squeeze()
print(f"\nNew tensor: {x_squeezed}")
```

```python
x_reshaped.shape
```

```python
x_reshaped.squeeze()
```

```python
x_reshaped.squeeze().shape
```

```python
#torch.unsqeeze-- adds a single dim to a target tensor at specific dim
print(f"Previous target: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

# Add an extra dimension with unsqueeze
x_unsqueezed=x_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```

```python
#torch.permute-- rearranges the dim of a target tensor in a specific order
x_original=torch.rand(size=(224,224,3)) #[height, width, color_channels]

# Permute the original tensor to rearrange the axis (or dim) order
x_permuted=x_original.permute(2,0,1) #[color_channels, height, width] # shifts axis 0->1, 1->2, 2->0

print(f"Previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")# colour trensors height and width
```

```python
x_original[0,0,0]= 728218
x_original[0,0,0],x_permuted[0,0,0]
```

## Reproducibility (trying to take random out of random)

In short how a neural network learns:
start with random numbers-> tensor operations -> update random numbers to try and make them better representations of the data -> again-> again-> again

To reduce the randomness in neural networks and pytorch comes the concept pf a "random seed"

Essentially what the random seed does is flavour the randomness

```python
torch.rand(3,3)
```

```python
import torch
# create two random tensors
random_tensor_A=torch.rand(3,4)
random_tensor_B=torch.rand(3,4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A==random_tensor_B)
```

```python
# lets make some random but reproducible tensors
import torch

#set the random seed
random_seed=42
torch.manual_seed(random_seed)
random_tensor_C=torch.rand(3,4)
torch.manual_seed(random_seed)
random_tensor_D=torch.rand(3,4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C==random_tensor_D)
```
